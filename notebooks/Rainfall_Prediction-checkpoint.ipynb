# Rainfall Prediction using Machine Learning
Binary classification: will it rain (yes/no) based on daily weather readings.

## 1. Import libraries

```python
import warnings
warnings.filterwarnings("ignore")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, roc_curve, confusion_matrix, classification_report)
sns.set_style("whitegrid")
```

## 2. Load data
Change the path if your file is elsewhere.

```python
df = pd.read_csv("Rainfall.csv")
df.columns = df.columns.str.strip()   # column names have stray spaces
print(df.shape)
df.head()
```

## 3. Explore the data

```python
df.info()
print("\nMissing values:\n", df.isnull().sum())
print("\nClass counts:\n", df["rainfall"].value_counts())
df.describe().T
```

```python
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
sns.countplot(x="rainfall", data=df, ax=ax[0]); ax[0].set_title("Rainfall class balance")
sns.heatmap(df.drop(columns=["day","rainfall"]).corr(), annot=True, fmt=".2f", cmap="coolwarm", ax=ax[1])
ax[1].set_title("Correlation matrix")
plt.tight_layout(); plt.show()
```

```python
feats = ["cloud", "sunshine", "humidity", "dewpoint", "pressure", "windspeed"]
fig, axes = plt.subplots(2, 3, figsize=(14, 7))
for a, f in zip(axes.ravel(), feats):
    sns.boxplot(x="rainfall", y=f, data=df, ax=a)
    a.set_title(f)
plt.tight_layout(); plt.show()
```

## 4. Preprocessing
- Encode `rainfall` yes/no to 1/0
- Drop `day` (it only cycles 1-31, no month/year given)
- Fill the 1 missing `winddirection` and `windspeed` with the median
- Drop `maxtemp` and `mintemp` (almost perfectly correlated with `temparature`)

```python
df["rainfall"] = df["rainfall"].str.strip().str.lower().map({"yes": 1, "no": 0})
df = df.drop(columns=["day"])
for c in ["winddirection", "windspeed"]:
    df[c] = df[c].fillna(df[c].median())
df = df.drop(columns=["maxtemp", "mintemp"])

X = df.drop(columns="rainfall")
y = df["rainfall"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)
print(X_train.shape, X_test.shape)
```

## 5. Train and compare models
Scaling is applied only to the distance/linear models via a Pipeline. `class_weight='balanced'` handles the yes/no imbalance.

```python
models = {
    "Logistic Regression": Pipeline([("s", StandardScaler()), ("m", LogisticRegression(max_iter=1000, class_weight="balanced"))]),
    "KNN": Pipeline([("s", StandardScaler()), ("m", KNeighborsClassifier(n_neighbors=7))]),
    "SVM (RBF)": Pipeline([("s", StandardScaler()), ("m", SVC(probability=True, class_weight="balanced", random_state=42))]),
    "Naive Bayes": GaussianNB(),
    "Decision Tree": DecisionTreeClassifier(max_depth=4, class_weight="balanced", random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=300, class_weight="balanced", random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42),
}
cv = StratifiedKFold(5, shuffle=True, random_state=42)
rows = []
for name, m in models.items():
    cv_acc = cross_val_score(m, X_train, y_train, cv=cv, scoring="accuracy").mean()
    m.fit(X_train, y_train)
    pred = m.predict(X_test); prob = m.predict_proba(X_test)[:, 1]
    rows.append([name, cv_acc, accuracy_score(y_test, pred), precision_score(y_test, pred),
                 recall_score(y_test, pred), f1_score(y_test, pred), roc_auc_score(y_test, prob)])
results = pd.DataFrame(rows, columns=["Model","CV Acc","Test Acc","Precision","Recall","F1","ROC-AUC"]).round(3)
results.sort_values("ROC-AUC", ascending=False).reset_index(drop=True)
```

```python
results.set_index("Model")[["Test Acc","F1","ROC-AUC"]].plot(kind="bar", figsize=(11,4))
plt.ylim(0.5, 1); plt.title("Model comparison"); plt.xticks(rotation=30, ha="right")
plt.tight_layout(); plt.show()
```

## 6. Hyperparameter tuning (Random Forest)

```python
grid = GridSearchCV(
    RandomForestClassifier(class_weight="balanced", random_state=42),
    {"n_estimators": [200, 400], "max_depth": [3, 5, 8, None], "min_samples_leaf": [1, 3, 5]},
    cv=cv, scoring="roc_auc", n_jobs=-1)
grid.fit(X_train, y_train)
print("Best params:", grid.best_params_)
print("Best CV ROC-AUC:", round(grid.best_score_, 3))
best = grid.best_estimator_
```

## 7. Final evaluation

```python
pred = best.predict(X_test); prob = best.predict_proba(X_test)[:, 1]
print("Accuracy:", round(accuracy_score(y_test, pred), 3))
print("ROC-AUC :", round(roc_auc_score(y_test, prob), 3))
print(classification_report(y_test, pred, target_names=["No Rain", "Rain"]))

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
sns.heatmap(confusion_matrix(y_test, pred), annot=True, fmt="d", cmap="Blues",
            xticklabels=["No Rain","Rain"], yticklabels=["No Rain","Rain"], ax=ax[0])
ax[0].set_xlabel("Predicted"); ax[0].set_ylabel("Actual"); ax[0].set_title("Confusion matrix")
fpr, tpr, _ = roc_curve(y_test, prob)
ax[1].plot(fpr, tpr, label=f"AUC = {roc_auc_score(y_test, prob):.2f}"); ax[1].plot([0,1],[0,1],"--")
ax[1].set_xlabel("False positive rate"); ax[1].set_ylabel("True positive rate"); ax[1].legend(); ax[1].set_title("ROC curve")
plt.tight_layout(); plt.show()
```

```python
imp = pd.Series(best.feature_importances_, index=X.columns).sort_values()
imp.plot(kind="barh", figsize=(7,4), title="Feature importance")
plt.tight_layout(); plt.show()
```

## 8. Predict for a new day

```python
new_day = pd.DataFrame([{
    "pressure": 1010, "temparature": 26, "dewpoint": 24, "humidity": 90,
    "cloud": 88, "sunshine": 0.5, "winddirection": 50, "windspeed": 20}])[X.columns]
p = best.predict_proba(new_day)[0, 1]
print(f"Chance of rain: {p:.0%} ->", "Rain" if p >= 0.5 else "No rain")
```
